# **2. Prepare The Data**

#### Transforming/organizing the data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

#### Nhap va kham pha du lieu

```python
retaild_df = pd.read_excel('/content/Retail Sales Data Project.xlsx')

retaild_df.describe()
```

```python
retaild_df.info()
```

```python
retaild_df.head()
```

```python
#Tong so row duplicate
retaild_df.duplicated().sum()
```

```python
#Loai bo duplicate
retaild_df.drop_duplicates()
```

```python
# Loai bo khoang trang cua cac cot co dang text
retaild_df['Customer ID'] = retaild_df['Customer ID'].str.strip()
retaild_df['Gender'] = retaild_df['Gender'].str.strip()
retaild_df['Product Category'] = retaild_df['Product Category'].str.strip()
```

```python
#Bo sung du lieu thieu bang mean va mode
retaild_df['Total Amount'] = retaild_df['Total Amount'].fillna(retaild_df['Total Amount'].mean())
retaild_df['Quantity'] = retaild_df['Quantity'].fillna(retaild_df['Quantity'].mode())
retaild_df['Price per Unit'] = retaild_df['Price per Unit'].fillna(retaild_df['Price per Unit'].mean())

# Loai bo du lieu thieu
text_columns = ['Customer ID', 'Gender', 'Product Category']
retaild_df = retaild_df.dropna(subset=text_columns, how="any")
```

```python
#Thuc hien dong bo du lieu text
retaild_df["Gender"] = retaild_df["Gender"].str.title()
retaild_df["Product Category"] = retaild_df["Product Category"].str.title()
retaild_df["Customer ID"] = retaild_df["Customer ID"].str.upper()
```

```python
retaild_df.head()
```

```python
#Thuc hien chuyen Date -> Datetime
retaild_df["Date"] = pd.to_datetime(retaild_df["Date"])
```

```python
retaild_df.info()
```

```python
# Tao cot phat sinh
retaild_df['year_month'] = retaild_df['Date'].dt.to_period('M')
retaild_df['month'] = retaild_df['Date'].dt.month_name()

bins = [0, 18, 35, 50, 65, 100]

labels = ['0-17', '18-34', '35-49', '50-64', '65+']

retaild_df['age_group'] = pd.cut(retaild_df['Age'], bins=bins, labels=labels, right=False)
```

```python
retaild_df.head()
```
